## 2. The question: decision, action, cost of a wrong call


*   *The Search Question: Which historical, observable search and engagement signals known prior to a decision window can reliably prioritize content items for a refresh queue?*
*   *The Unit of Analysis (Grain): A single pseudonymized content item (content_hash_id) belonging to a specific client (client_hash_id).*
*   *The Output: A ranked opportunity probability score paired with explicit structural reason codes.*
*   *The Action Taken: A human content reviewer surfaces the top-$K$ recommendations from the queue to prioritize where to spend their limited editing and optimization hours first.*

*   *The Cost of a Wrong Recommendation: A false positive (ranking an un-fixable page highly) wastes high-cost human editorial hours on dead traffic. A false negative (missing a major decline) leaves critical client traffic unprotected and unmonetized.*
*   *Why ML/Data Helps: Fixed rule baselines cannot dynamically weight non-linear combinations of signals (such as the steep drop-off in CTR across different ranking position tiers vs. word count constraints).*
